# 01 - Load the PhysioNet 2019 sepsis data

Reads the ~40,000 patient files (`.psv`, one file per patient, one row per ICU hour) from the two hospital systems, adds `patient_id` (file name) and `hospital` (A/B), stacks everything into one table and saves it as a pickle so the other notebooks start in seconds.

Expected input: the folders `data/raw/training_setA` and `data/raw/training_setB` containing the `.psv` files (sub-folders are fine, see the README). Run once.

```python
import pandas as pd
from pathlib import Path

RAW_DIR = Path("../data/raw")              # contains training_setA/ and training_setB/ (.psv files)
PROCESSED_DIR = Path("../data/processed")  # the combined pickle is written here
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
```

```python
sources = {"A": RAW_DIR / "training_setA", "B": RAW_DIR / "training_setB"}
data = []
for hospital, path in sources.items():
    for file_path in sorted(path.rglob("*.psv")):  # works with nested sub-folders
        file = pd.read_csv(file_path, sep="|")
        file["patient_id"] = file_path.stem
        file["hospital"] = hospital
        data.append(file)
assert data, "No .psv files found in data/raw/training_setA and training_setB"
data = pd.concat(data, ignore_index=True)
```

```python
data.to_pickle(PROCESSED_DIR / "DataFrame.pkl")
```
